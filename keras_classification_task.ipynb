```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
df = pd.DataFrame(cancer['data'], columns=[
    'mean radius', 'mean texture', 'mean perimeter', 'mean area',
    'mean smoothness', 'mean compactness', 'mean concavity',
    'mean concave points', 'mean symmetry', 'mean fractal dimension',
    'radius error', 'texture error', 'perimeter error', 'area error',
    'smoothness error', 'compactness error', 'concavity error',
    'concave points error', 'symmetry error', 'fractal dimension error',
    'worst radius', 'worst texture', 'worst perimeter', 'worst area',
    'worst smoothness', 'worst compactness', 'worst concavity',
    'worst concave points', 'worst symmetry', 'worst fractal dimension'
])
df['benign_0__mal_1'] = cancer['target']
df.info()
```

```python
df.describe()
```

## EDA

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(x='benign_0__mal_1', data=df)
plt.title('Class Distribution: Benign (0) vs Malignant (1)')
plt.show()
```

```python
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
# Based on 'benign_0__mal_1'
X = df.drop('benign_0__mal_1', axis=1).values
y = df['benign_0__mal_1'].values
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=101
)
```


## Scaling Data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Creating the Model

                  
    

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation
```

```python
model = Sequential([
    Dense(30, activation='relu', input_shape=(30,)),
    Dense(15, activation='relu'),
    Dense(1,  activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

## Training the Model

### Example One: Choosing too many epochs and overfitting!

```python
model.fit(X_train, y_train, epochs=600, verbose=0)
```

```python
# model.history.history
model_loss = pd.DataFrame(model.history.history)
model_loss.head()
```

```python
# model_loss
model_loss.plot()
plt.title('Model Training Loss & Accuracy')
plt.xlabel('Epoch')
plt.show()
```

```python
model_loss.plot()
```

## Example Two: Early Stopping

We obviously trained too much! Let's use early stopping to track the val_loss and stop training once it begins increasing too much!

```python
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Dense, Activation, Dropout
```

```python
early_stop = EarlyStopping(monitor='val_loss', patience=25, restore_best_weights=True)
```

```python
model2 = Sequential([
    Dense(30, activation='relu', input_shape=(30,)),
    Dropout(0.5),
    Dense(15, activation='relu'),
    Dropout(0.5),
    Dense(1,  activation='sigmoid')
])
model2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model2.fit(
    X_train, y_train,
    epochs=600,
    validation_data=(X_test, y_test),
    callbacks=[early_stop],
    verbose=0
)
```

## Example Three: Adding in DropOut Layers

```python
from tensorflow.keras.layers import Dropout
```

```python
model_loss2 = pd.DataFrame(model2.history.history)
model_loss2[['loss','val_loss']].plot()
plt.title('Loss: With EarlyStopping + Dropout')
plt.show()
```

```python
model_loss2[['accuracy','val_accuracy']].plot()
plt.title('Accuracy: With EarlyStopping + Dropout')
plt.show()
```

# Model Evaluation

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
predictions = (model2.predict(X_test) > 0.5).astype(int).flatten()
print(classification_report(y_test, predictions))
```

```python
print(confusion_matrix(y_test, predictions))
```

```python
# Save model and scaler weights
import pickle

model2.save('model_weights.h5')

with open('scaler_weights.pkl', 'wb') as f:
    pickle.dump(scaler, f)

# Save model as pkl
model_weights = {
    'weights': [w.numpy().tolist() for w in model2.weights],
    'config': model2.get_config()
}
with open('model_weights.pkl', 'wb') as f:
    pickle.dump(model_weights, f)

print('model_weights.pkl and scaler_weights.pkl saved successfully!')
```
